```python
price=[250,400,99,1500]
total=0
for i in price:
    total=total+i
if total>1000:
    print("you get a 10% discount!")
    dis_price=total-(total*(10/100))
    print(dis_price)
else:
    print(total)
```

```python
num=[2,4,6,8,-1,-3,-5,-7,0,9]
for i in num:
    if i>0:
        print(i,"positive")
    elif i<0:
        print(i,"negative")
    else:
        print(i,"zero")
```
