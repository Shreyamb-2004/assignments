```python
price=[250,400,99,1500]
total=0
for i in price:
    total=total+i
if total>1000:
    print("you get a 10% discount!")
    dis_price=total-(total*(10/100))
    print(dis_price)
else:
    print(total)
```

```python
num=[2,4,6,8,-1,-3,-5,-7,0,9]
for i in num:
    if i>0:
        print(i,"positive")
    elif i<0:
        print(i,"negative")
    else:
        print(i,"zero")
```

```python
name={"riya":96,"tina":83,"minnu":64,"nisha":56,"yash":87}
print(name)
student=input("enter name:")
if student in name:
    print("marks:",name[student])
else:
    print("not found")
highest=max(name.values())
for keys,values in name.items():
    if values==highest:
        print(f"{keys} has scored highest marks of {highest}")
```

```python
class student:
    def __init__(self,name,age,course):
        self.name=name
        self.age=age
        self.course=course
    def display(self):
        print(f"name:{self.name},age:{self.age},course:{self.course}")
s1=student("riya",20,"BCA")
s2=student("yash",22,"BBA")
s1.display()
s2.display()


```
